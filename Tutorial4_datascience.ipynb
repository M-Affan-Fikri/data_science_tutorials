#Question 1: Identify the Problem
Consider this dataset:

Student	Math	Science	English

Ali  	75  	80  	72

Sara  	88  	91  	85

John	  65	  70  	68

1. Is this data in tidy format?
2. If not, explain which tidy-data rule is violated.
3. Using Python/pandas, reshape the dataset so that it has the columns:
   - Student
   - Subject
   - Score

answer:
1. No, this dataset is not in tidy format
2. rule violated; "Each variable forms a column and each observation forms a row.

```python
#3.
import pandas as pd
#create original dataframe
data_q1 = {
    'Student': ['Ali', 'Sara', 'John'],
    'Math': [75, 88, 65],
    'Science': [80, 91, 70],
    'English': [72, 85, 68]
}
df_q1 = pd.DataFrame(data_q1)
#reshape using melt
tidy_df_q1 = df_q1.melt(
    id_vars=['Student'],
    var_name='Subject',
    value_name='Score'
)

print(tidy_df_q1)
```

answer:

1\. The quarter columns (Q1_Sales, Q2_Sales, Q3_Sales, Q4_Sales) represent values of a time variable rather than distinct column header. Ech row contains multiple observations for the same entity.


```python
#2
import pandas as pd
#create original dataframe
data_q2={
    'Country':['Malaysia', 'Thailand', 'Indonesia'],
    'Q1_Sales':[12000,10000,14000],
    'Q2_Sales':[15000,11000,12500],
    'Q3_Sales':[14000,13000,15000],
    'Q4_Sales':[18000,16000,17000]
}
df_q2 = pd.DataFrame(data_q2)
#reshape using melt
tidy_df_q2 = df_q2.melt(
    id_vars=['Country'],
    var_name='Quarter',
    value_name='Sales'
)
print(tidy_df_q2)
```

#Question 3


```python
import pandas as pd
import matplotlib.pyplot as plt

#load dataset
data = {    "Student": ["A", "B", "C", "D", "E", "F"],
            "Study_Hours": [2, 5, 3, 7, 4, 8],
            "Score": [55, 72, 60, 85, 68, 90]
}
df_q3 = pd.DataFrame(data)
#1
avg_study_hours = df_q3["Study_Hours"].mean()
print(f"1. Average study time; {avg_study_hours} hours")
#2
avg_score = df_q3["Score"].mean()
print(f"2. Average score; {avg_score}")
#3
top_student_score = df_q3.loc[df_q3["Score"].idxmax(), "Student"]
print(f"3. Student with the highest score: {top_student_score}")
#4
top_student_hours = df_q3.loc[df_q3["Study_Hours"].idxmax(), "Student"]
print(f"4. Student who studied the most: {top_student_hours}")
#5
print("5. Association; Yes, as study hourse increase, scores also increase.")
#6
plt.figure(figsize=(6,4))
plt.scatter(df_q3["Study_Hours"],df_q3["Score"],color='blue',s=100)
plt.title("Study Hours vs Score")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.grid(True,linestyle='--',alpha=0.6)
plt.show()


```
