```python
ones = np.ones((5,5))
zeros = np.zeros((2,10))
print(ones)
print(zeros)
others = np.full((2,4),2.8)
print(others)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product


```python
#1. Create 2 random arrays of size n x p and p x m
n,p,m = 2,1,2
array_a = np.random.rand(n,p)
array_b = np.random.rand(p,m)
print("Array A ({}x{}): ".format(n,p))
print(array_a)
print("Array B ({}x{}): ".format(p,m))
print(array_b)

#2. Compute dot product
dot = np.dot(array_a,array_b)

print("\nDot product result ({}x{}): ".format(2,2))
print(dot)
```
